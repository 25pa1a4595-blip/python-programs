```python
def pen(num):
  ori=num
  rev=0
  while(num!=0):
    rem=num%10
    rev=rev*10 + rem
    num=num//10
  if(rev==ori):
    print("palindrome")
  else:
    print("not a palindrome")
pen(121)
```

```python
def rev_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1): #hello
     rev=rev+s[i]
  return rev
print(rev_str("john"))
```

```python
def valve(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
sum,count,avg=valve(1,2,3,4,5)
print(f"sum is {sum},count is {count},avg is {avg}")
```
