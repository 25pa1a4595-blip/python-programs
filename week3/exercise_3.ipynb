```python
#1.print 1 to n odd numbers
n=int(input())
for i in range(1,n+1,2):
  print(i,end=" ")
```

```python
#2. print multiplication table of a given number 2 * 1 = 2
n=int(input())
for i in range(1,21):
  print(f"{n} * {i} = {n*i}")
```

```python
#3. check whether thr given nums is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("it is not a prime")
    break
else:
    print("yes it is prime")
```

```python
#4. print prime number between given 2 no's
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
   if(i%j==0):
    break
  else:
    print(i,end=" ")
```

```python
#5. print the sum of the digits of a given number
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
  print(sum)
```
