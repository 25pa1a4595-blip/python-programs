```python
#1.indexing
L1=[1,2,3,55,66,88,99,0,"hello","name"]
L2=[100,200,400]
print(L1+L2) #concatenation
print(L2*3) #repetition
print(L1[5]) #indexing , prints element at 5th index
print(100 in L2) #check 100 present in list L2 or not
print(L1[2:7]) #slicing,print element in reverse order
print(L1[8::-7])
print(L1[-7:-1])
for i in L2:
  print(i) #iteration throuhgh a list
```
